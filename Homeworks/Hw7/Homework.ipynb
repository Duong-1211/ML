# Các thư viện sử dụng 

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.metrics.pairwise import euclidean_distances
%matplotlib inline
```

# Chuẩn bị dữ liệu 
- Sinh dữ liệu ngẫu nhiên n_samples = 100 tương đương 100 điểm 
    - random_state: biến cố định hàm random - để các điểm sinh ngẫu nhiên giống nhau giữa các máy tính
    
- Mỗi điểm dữ liệu có 2 chiều 

```python
n_samples = 100
random_state = 170
center_points = [[1, 1], [-1, -1], [1, -1]] # sinh ngẫu nhiên các điểm xung quanh vị trí tâm cố định 
#center_points = 3                           # tâm cụm được chọn ngẫu nhiên

X, y = make_blobs(n_samples=n_samples, random_state=random_state, centers=center_points, cluster_std=0.6)
print("Số chiều dữ liệu: ", X.shape, y.shape)
print("5 điểm dữ liệu đầu tiên: \n", X[:5])
```

```python
plt.figure(figsize=(12, 12))
plt.subplot(221)
plt.scatter(X[:, 0], X[:, 1], c='orange') # c là tham số chọn màu sắc, có thể truyền vào string hoặc số id 1,2,3 ...
plt.title("Các điểm dữ liệu trước khi phân cụm. Số lượng: {}".format(n_samples))
for point in center_points:
    plt.plot(point[0], point[1], 'ko', markersize=5, markeredgewidth=3)
plt.show()
```

# Tự xây dựng giải thuật K-means:

Viết code cho giải thuật K-means tại mục này

```python
def KMeans_init(X, n_clusters):
    n_samples, n_features = X.shape
    # Create initial empty array of centroids
    init_centroid = np.empty((n_clusters, n_features), dtype=X.dtype)
    avg_point = sum(X)/len(X)

    # Calculate the distance of all points to the midpoint
    dis_to_avg = euclidean_distances([avg_point], X, squared = True)
    center_id = np.argmax(dis_to_avg)
    init_centroid[0] = X[center_id] #Choose a centroid which is far the most from the midpoint

    for i in range(1, n_clusters):
        distance = euclidean_distances(init_centroid[:i], X, squared = True)
        min_distance = np.min(distance, axis=0)
        next_center_id = np.argmax(min_distance)
        init_centroid[i] = X[next_center_id]
    init_centroid = init_centroid.tolist()
    return init_centroid
k_clusters = 3
init_centroid = KMeans_init(X, n_clusters=k_clusters)
init_centroid
```

```python
def KMeans_model(X, n_clusters):
    n_samples = len(X)
    new_centroid = KMeans_init(X, n_clusters)
    best_centroid = None
    label = np.empty(n_samples, dtype=int)

    while best_centroid != new_centroid:
        best_centroid = new_centroid.copy()
        
        distance = euclidean_distances(X, best_centroid, squared = True)

        for i in range(n_samples):
            label[i] = np.argmin(distance[i])

        for j in range(n_clusters):
            label_j = np.where(label == j)[0]
            cluster_j = X[label_j]
            new_centroid[j] = sum(cluster_j)/len(cluster_j)
            new_centroid[j] = new_centroid[j].tolist()
    best_centroid = np.asarray(best_centroid)
    return best_centroid, label
```

# Kiểm tra giải thuật K-means tự viết cho dữ liệu sinh ngẫu nhiên

- Áp dụng giải thuật K-means tự viết cho tập dữ liệu đã sinh ngẫu nhiên ở trên
- Quan sát kết quả và so sánh với giải thuật của sklearn

```python
best_centroid, label = KMeans_model(X, n_clusters=3)
print(label)
print(best_centroid)
plt.figure(figsize=(12, 12))
plt.subplot(222)
plt.scatter(X[:, 0], X[:, 1], c=label)
for point in best_centroid:
    plt.plot(point[0], point[1], 'ko', markersize=5, markeredgewidth=3)
plt.title("KMeans. Số cụm: {}".format(k_clusters))
plt.show()
```

# Ứng dụng K-means tự viết vào nén ảnh 

## Thư viện sử dụng - hỗ trợ hình ảnh 

```python
from skimage import io
from sklearn.cluster import KMeans
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as image
from IPython.core.display import Image, display
```

## Đọc dữ liệu hình ảnh
- Mỗi điểm ảnh là 1 mẫu quan sát 
- Phân cụm tập dữ liệu (tập các điểm ảnh) về k nhãn

```python
path_img = 'bird_small.png'
display(Image(path_img, width=250, unconfined=True))
img = io.imread(path_img)
data_img = (img / 255.0).reshape(-1,img.shape[2]) # chuyển ma trận 128x128x3 về mảng 2 chiều 
img_shape = img.shape                    

print("Số chiều của dữ liệu hình ảnh: ", data_img.shape)
print("Tổng số điểm ảnh là: ", data_img.shape[0])
print("Mỗi điểm ảnh có số chiều = ", data_img.shape[1])
```

## Nén ảnh bằng giải thuật K-means tự viết

- Tạo file nén ảnh bằng giải thuật K-means tự viết
- Hiển thị kết quả của giải thuật tự viết và giải thuật của sklearn để so sánh


Ví dụ

```python
n_colors = 16
best_centroid, label = KMeans_model(data_img, n_clusters=n_colors)
img = best_centroid[label]

img = np.reshape(img, img_shape)

image.imsave('img.png', img)
```

```python
n_clusters = 16
kmeans = KMeans(n_clusters=n_clusters, random_state=0).fit(data_img)
img2 = kmeans.cluster_centers_[kmeans.labels_]

img2 = np.reshape(img2, img_shape)

image.imsave('img2.png', img2)
```

```python
print('Ảnh nén bằng K-means tự viết')
display(Image('img.png', width=250, unconfined=True)) #kết quả tự cài đặt
print('Ảnh nén bằng K-means của thư viện ')
display(Image('img2.png', width=250, unconfined=True)) #kết quả của thư viện
print('Ảnh gốc')
display(Image(path_img, width=250, unconfined=True))
```
